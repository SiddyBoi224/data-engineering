### `docker ps` vs `docker ps -a`

`docker ps` shows only currently running containers.

`docker ps -a` shows all containers, including stopped/exited containers.

Each execution of `docker run` creates a new container.

A container can be stopped/exited while still existing on the system.

### Docker Images

`docker images` lists the Docker images available locally.

An image is a template used to create containers.

The same image can be used to create multiple containers.

For example, the `hello-world:latest` image has been used to create multiple containers on this machine.

The key concept
Think of an image as a template:
Docker Image
│
├──► Container A
├──► Container B
└──► Container C
