### Docker Volumes

A Docker volume stores data outside the container's writable filesystem.

This means the data can survive even if the container is removed.

Example:

`docker run --name volume-test -v docker-data:/data ubuntu sh -c "echo 'Hello Docker' > /data/test.txt"`

- `-v docker-data:/data` → mounts the `docker-data` volume at `/data`
- Data written to `/data` is stored in the Docker volume
- The volume remains after the container is removed

The same volume can be mounted into another container:

`docker run --rm -v docker-data:/data ubuntu cat /data/test.txt`

### Key Point

**Container → temporary**

**Volume → persistent data**

````python
!docker run --name volume-test -v docker-data:/data ubuntu sh -c "echo 'Hello Docker' > /data/test.txt"
````

````python
!docker rm volume-test
````

````python
!docker run --rm -v docker-data:/data ubuntu cat /data/test.txt
````

### Bind Mounts

A bind mount maps a directory or file from the host machine directly into a container.

Example:

`docker run --rm -v "$(pwd)":/data ubuntu ls /data`

- `$(pwd)` → current directory on the host
- `/data` → directory inside the container
- Changes on the host are visible inside the container.

### Volume vs Bind Mount

**Volume**
- Managed by Docker
- Good for persistent application data

**Bind mount**
- Uses a specific host directory
- Useful for development and sharing files with containers

````python
!docker run --rm -v "$(pwd)":/data ubuntu ls /data
````
