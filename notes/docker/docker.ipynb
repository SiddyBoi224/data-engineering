### `docker ps` vs `docker ps -a`

`docker ps` shows only currently running containers.

`docker ps -a` shows all containers, including stopped/exited containers.

Each execution of `docker run` creates a new container.

A container can be stopped/exited while still existing on the system.

```python
!docker images
```

### Docker Images

`docker images` lists the Docker images available locally.

An image is a template used to create containers.

The same image can be used to create multiple containers.

For example, the `hello-world:latest` image has been used to create multiple containers on this machine.

The key concept
Think of an image as a template:
Docker Image
│
├──► Container A
├──► Container B
└──► Container C

### Pull an Image

`docker pull` downloads an image from Docker Hub and stores it locally.

Example:

```python
!docker pull ubuntu
```

### Run a Container

`docker run` creates and starts a new container from an image.

If the container's main process finishes, the container stops but still exists.

Example:

```python
!docker run ubuntu
```

```python
!docker ps -a
```

### Interactive Container

`docker run -it ubuntu bash` starts an Ubuntu container with an interactive terminal.

- `-i` → interactive input
- `-t` → terminal
- `bash` → starts the Bash shell inside the container

Use `exit` to leave the container.

```python
!docker run -it ubuntu bash
```

### Start a Stopped Container

`docker start` starts an existing stopped container.

It does not create a new container.

Example:

```python
!docker start relaxed_heisenberg
```

```python
!docker ps -a
```

### Naming a Container

By default, Docker generates a random container name.

Use `--name` to assign your own name:

`docker run --name <name> <image> <command>`

```python
!docker run -d --name ubuntu-test ubuntu sleep 300
```

```python
!docker ps
```

### Stop a Container

`docker stop` gracefully stops a running container.

The container is not deleted and can be started again using `docker start`.

```python
!docker stop ubuntu-test
```

### Restart a Container

`docker restart` stops a running container and starts it again.

It works on an existing container and does not create a new one.

```python
!docker restart ubuntu-test
```

```python
!docker ps
```

```python
!docker stop ubuntu-test
```

### Remove a Container

`docker rm` permanently removes a stopped container.

It does not remove the Docker image.

`docker stop` → stops the container
`docker rm` → removes the container

```python
!docker rm ubuntu-test
```

```python
!docker ps -a
```
